```python
#1. Create a numpy array of size 2 x 2
import numpy as np
array_2x2 = np.array ([
    [1,2],
     [3,4]
])
print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4
random_3x4 = np.random.rand(3,4)
print("This is my random 3x4 array")
print(random_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3,3))
zeroes_array = np.zeros((3,3))

print("Array of ones:")
print(ones_array)

print("\nArray of zeroes:")
print(zeroes_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it
random_6x3 = np.random.rand(6,3)
transposed_array = random_6x3.T
print("This is my random 6x3 array")
print(random_6x3)
print("\nThis is the transposed array")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
random_6x3 = np.random.rand(6,3)
random_3x4 = np.random.rand(3,4)
dot_product = np.dot(random_6x3, random_3x4)
print("This is my random 6x3 array")
print(random_6x3)
print("\nThis is my random 3x4 array")
print(random_3x4)
print("\nThis is the dot product of the two arrays")
print(dot_product)
```

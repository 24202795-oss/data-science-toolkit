```python
import pandas as pd

student_data = {
    "Student": ["Ali", "Sara", "John"],
    "Math": [75, 88, 65],
    "Science": [80, 91, 70],
    "English": [72, 85, 68]}


students_df=pd.DataFrame(student_data)

print("Original student data:")
display (students_df)

tidy_students=pd.melt(
    students_df,
    id_vars=["Student"],
    var_name="Subject",
    value_name="Score")

print("\nTidy student data:")
display (tidy_students)
```

```python
sales_data = {
    "Country": ["Malaysia", "Thailand", "Indonesia"],
    "Q1_Sales": [12000, 10000, 14000],
    "Q2_Sales": [15000, 11000, 12500],
    "Q3_Sales": [14000, 13000, 15000],
    "Q4_Sales": [18000, 16000, 17000]
}


sales_df=pd.DataFrame(sales_data)

print("Original sales data:")
display(sales_df)


tidy_sales=pd.melt(
    sales_df,
    id_vars=["Country"],
    var_name="Quarter",
    value_name="Sales"
)


tidy_sales["Quarter"]= tidy_sales["Quarter"].str.replace("_Sales", "", regex=False)
tidy_sales=tidy_sales.sort_values (["Country", "Quarter"]).reset_index(drop=True)

print("Tidy sales data:")
display(tidy_sales)
```

```python
#Enter the study-hours and exam-score data.

study_data={
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 5, 3, 7, 4, 8],
    "Score": [55, 72, 60, 85, 68, 90]
    }

study_df=pd.DataFrame (study_data)
display(study_df)

#Q1: mean() finds the average of the numbers in a column.
average_hours = study_df["Study_Hours"].mean()
print (f"1. Average study time: {average_hours:.2f} hours")

#Q2: Find the average exam score.
average_score=study_df["Score"].mean()
print(f"2. Average score: {average_score:.2f}")

#Q3:If you sort the students by highest score, then the first row will naturally have the highest value
#Sort students by score (highest first)

sorted_df=study_df.sort_values("Score", ascending=False)
#Display the student with the highest score
print ("3. Student with the highest score:")
print(sorted_df.head(1))

#Q4: Find the row with the largest number of study hours.
#An alternative method slightly harder
#Find the student with the largest study hours
largest_hours=study_df.loc[study_df["Study_Hours"].idxmax()]

#Print the student's name and study time
print("4. Largest study hours: Student",largest_hours["Student"],"(",largest_hours["Study_Hours"],"hours)",largest_hours["Score"])
print("5. Yes, studying more hours appear to be associated with a higher score")
```

```python
import matplotlib.pyplot as plt
#Create a figure (the area where our graph will appear).
#figsize (7, 5) sets the width to 7 inches and height to 5 inches.

plt.figure(figsize=(7, 5))

#Create a scatter plot (a graph made of dots).
#Each dot represents one student.
#Study_Hours gives the position on the X-axis (horizontal).
#Score gives the position on the Y-axis (vertical).
#s-80 controls the size of the dots.
plt.scatter(study_df["Study_Hours"], study_df ["Score"], s=80)

#Add a title at the top of the graph.
plt.title("Study Hours vs Student Scores")

#Add a label to the horizontal (X) axis.
plt.xlabel("Study Hours")

#Add a label to the vertical (Y) axis.
plt.ylabel("Score")

# Display grid lines to help us read the graph.
#alpha-0.3 makes the grid lines lighter.
#The alpha value ranges from e (invisible) to 1 (solid).

plt.grid(True, alpha=0.3)

#Display the completed graph.
plt.show()
```



6. & 8. How many columns have null values, missing values? If yes, which column, which rows



7. Which column is the feature, which is the target output?


```python
target = "Present"
feature = df.drop(columns=[target])
print("Target: ",target)
print("Feature columns: ",feature.columns.tolist())
print("Target values: ", df[target].unique())

```

9. What kind of question can be asked, should we ask. Does the dataset suits the question we are asking?


```python
print("What is the average daily attendance rate across all schools?")

print("\n---Answers---")
total_present = df['Present'].sum()
total_enrolled = df['Enrolled'].sum()
average_attendance_rate = (total_present / total_enrolled) * 100

print("Total Students Enrolled: ",total_enrolled)
print("Total Students Present: ",total_present)
print(f"Average Daily Attendance Rate: {average_attendance_rate:.2f}%")

print("\nYes! The dataset is perfectly suited for these questions. ")
print("It has a robust sample size of 277,153 daily records, ")
print("covers distinct date periods for the 2018-2019 school year, ")
print("identifies individual schools uniquely via School DBN, ")
print("and contains clear numeric metrics (Enrolled, Absent, Present, Released).")
```
